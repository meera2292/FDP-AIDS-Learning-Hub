```python
import numpy as np

# Data
X = np.array([1, 2, 3, 4])
y = np.array([2, 4, 6, 8])
```

```python
len(X)
```

```python
np.ones(4)
```

```python
np.c_[np.ones(len(X)), X]
```

```python
X_b = np.c_[np.ones(len(X)), X]
print("X with bias:")
print(X_b)
```

X_b.T

1  1  1  1
1  2  3  4

X_b.T @ X_b

It multiplies the transpose of X by X.

np.linalg.inv(X_b.T @ X_b)

(XTX)−1

“NumPy is doing the matrix inverse calculation for us.”

```python
theta = np.linalg.inv(X_b.T @ X_b) @ X_b.T @ y
print("Theta:", theta)
```

```python
b = theta[0]
w = theta[1]

print("w =", w)
print("b =", b)
```

```python
x_new = 5
prediction = w * x_new + b
print("Prediction for x = 5:", prediction)
```
