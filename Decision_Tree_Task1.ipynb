```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier, plot_tree
import matplotlib.pyplot as plt

```

```python
# Dataset
data = {
    "Weather": ["Sunny", "Sunny", "Rainy", "Rainy", "Cloudy", "Cloudy"],
    "Wind": ["Weak", "Strong", "Weak", "Strong", "Weak", "Strong"],
    "Play_Cricket": ["Yes", "No", "No", "No", "Yes", "Yes"]
}

df = pd.DataFrame(data)
```

```python
# Convert categorical data into numbers
df["Weather"] = df["Weather"].map({
    "Sunny": 0,
    "Rainy": 1,
    "Cloudy": 2
})

df["Wind"] = df["Wind"].map({
    "Weak": 0,
    "Strong": 1
})

df["Play_Cricket"] = df["Play_Cricket"].map({
    "No": 0,
    "Yes": 1
})
```

```python
# Input features and target
X = df[["Weather", "Wind"]]
y = df["Play_Cricket"]
```

```python

# Create and train Decision Tree
model = DecisionTreeClassifier(criterion="entropy", random_state=0)
model.fit(X, y)
```

```python
# Visualize the Decision Tree
plt.figure(figsize=(12, 6))

plot_tree(
    model,
    feature_names=["Weather", "Wind"],
    class_names=["Don't Play", "Play"],
    filled=True,
    rounded=True
)

plt.show()
```

```python

# Predict for a new example
# Sunny = 0, Weak Wind = 0
prediction = model.predict([[0, 0]])

if prediction[0] == 1:
    print("Prediction: Play Cricket")
else:
    print("Prediction: Don't Play Cricket")
```
