```python
import numpy as np
from collections import Counter

# Sample data: 2 classes (0 and 1)
X_train = np.array([
    [1, 2], [2, 3], [3, 1],     # Class 0
    [6, 5], [7, 7], [8, 6]      # Class 1
])
y_train = np.array([0, 0, 0, 1, 1, 1])

# New point to classify
x_new = np.array([5, 5])

# Function to calculate Euclidean distance
def euclidean_distance(a, b):
    return np.sqrt(np.sum((a - b)**2))

# KNN implementation
def knn_predict(x_new, X_train, y_train, k=3):
    distances = []

    # Step 1: Calculate all distances
    for i in range(len(X_train)):
        dist = euclidean_distance(x_new, X_train[i])
        distances.append((dist, y_train[i]))

    # Step 2: Sort by distance and pick k closest
    distances.sort()
    k_nearest_labels = [label for (_, label) in distances[:k]]

    # Step 3: Majority vote
    prediction = Counter(k_nearest_labels).most_common(1)[0][0]
    return prediction

# Predict class of new point
print("Predicted class:", knn_predict(x_new, X_train, y_train, k=3))
```
