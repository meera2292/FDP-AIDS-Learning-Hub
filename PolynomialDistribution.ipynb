```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x=np.array([[2,3,1],[1,2,1],[2,1,1]])
y=np.array([1,2,3])
t=x.transpose()
m=np.dot(x,t)
m
```

```python
invers=np.linalg.inv(m)
invers
```

```python
weights=np.dot(t,y)
finalweights=np.dot(weights,invers)
finalweights
```

```python

m= finalweights[0]
n=finalweights[1]
c=finalweights[2]
print(m,n,c)
y1 =np.dot(m,x)**2+np.dot(n,x)+c
plt.plot(x,y1,color='red')
x2=np.array([3,2,1])
y2=np.array([1,2,3])
plt.scatter(x2,y2)
plt.show()
```
